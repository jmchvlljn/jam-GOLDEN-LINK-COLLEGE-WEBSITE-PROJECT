```python
# Import required libraries
import pandas as pd

# Load the dataset
file_path = '/content/sample_data/Artistic_Skill_Classification_Survey_3.csv'  # Update the file path if necessary
df = pd.read_csv(file_path)

# Display the first few rows to inspect the dataset
df.head()

# Step 1: Remove leading and trailing spaces from column names
df.columns = df.columns.str.strip()

# Step 2: Handle missing or NaN values

df_cleaned = df.fillna('Unknown')

# Step 3: Remove any rows with unwanted characters or fix incorrect ones
# For example, correcting emails or unwanted characters
df_cleaned['Identity'] = df_cleaned['Identity'].str.replace(r'[^a-zA-Z0-9@.\s,]', '', regex=True)

# Step 4: Handling duplicates
df_cleaned = df_cleaned.drop_duplicates()

# Step 5: Clean and format any specific columns (example: standardizing career goals, expertise levels)
df_cleaned['CareerGoals'] = df_cleaned['CareerGoals'].str.strip()
df_cleaned['ExpertiseLevel'] = df_cleaned['ExpertiseLevel'].str.strip()

# Step 6: Convert data types if necessary (e.g., 'Timestamp' column to datetime)
df_cleaned['Timestamp'] = pd.to_datetime(df_cleaned['Timestamp'], errors='coerce')

# Display the cleaned DataFrame to inspect the results
df_cleaned.head()

# Optionally, save the cleaned data back to a CSV
df_cleaned.to_csv('/content/sample_data/cleaned_Artistic_Skill_Classification_Survey_3.csv', index=False)

print("Data cleaning completed and saved to '/content/sample_data/cleaned_Artistic_Skill_Classification_Survey_3.csv'.")
```

```python
import pandas as pd
from collections import Counter

# Load the cleaned data
df_cleaned = pd.read_csv('/content/sample_data/cleaned_Artistic_Skill_Classification_Survey_3.csv')

# Step 1: Define function to extract skills from a comma-separated string
def extract_skills(column):
    skills = []
    for skill_str in column.dropna():
        skills.extend([skill.strip() for skill in skill_str.split(',')])
    return skills

# Step 2: Extract skills for Digital and Traditional expertise
df_cleaned['DigitalExpertise_List'] = df_cleaned['DigitalExpertise'].apply(lambda x: extract_skills(pd.Series([x])))
df_cleaned['TraditionalExpertise_List'] = df_cleaned['TraditionalExpertise'].apply(lambda x: extract_skills(pd.Series([x])))

# Step 3: Function to extract top skills for each category (Foundational, Technical, Conceptual)
def extract_top_skills(group, skills_column, category_name):
    all_skills = []

    # Extract skills from the relevant expertise column
    skills = [skill for sublist in group[skills_column].dropna() for skill in sublist]
    skills_count = Counter(skills)

    # Return the top 3 most common skills in the category
    top_skills = skills_count.most_common(3)  # You can adjust this to top 5 or any number
    return [skill for skill, count in top_skills]

# Step 4: Function to display the expertise details for each type (Digital/Traditional)
def display_expertise_results(expertise_category, expertise_column, category_name):
    unique_expertise = df_cleaned[expertise_column].dropna().explode().unique()

    for expertise in unique_expertise:
        expertise_df = df_cleaned[df_cleaned[expertise_column].apply(lambda x: expertise in x if isinstance(x, list) else False)]

        top_foundational = extract_top_skills(expertise_df, expertise_column, "Foundational Skills")
        top_technical = extract_top_skills(expertise_df, expertise_column, "Technical Skills")
        top_conceptual = extract_top_skills(expertise_df, expertise_column, "Conceptual Skills")

        # Output the results
        print(f"\n{expertise_category} {expertise}")
        print(f"Top 5 Foundational Skills")
        for skill in top_foundational[:5]:
            print(f"- {skill}")

        print(f"\nTop 3 Technical Skills")
        for skill in top_technical[:3]:
            print(f"- {skill}")

        print(f"\nTop 3 Conceptual Skills")
        for skill in top_conceptual[:3]:
            print(f"- {skill}")

        print("\n" + "-"*50)

# Step 5: Display results for Digital Expertise
display_expertise_results("Digital", 'DigitalExpertise_List', "Digital Expertise")

# Step 6: Display results for Traditional Expertise
display_expertise_results("Traditional", 'TraditionalExpertise_List', "Traditional Expertise")
```
